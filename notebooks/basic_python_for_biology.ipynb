# Basic Python for Biology


```python
dna_sequence = "ATGCGATACGCTTGA"
print(dna_sequence)
```

```python
count_A = dna_sequence.count("A")
count_T = dna_sequence.count("T")
count_G = dna_sequence.count("G")
count_C = dna_sequence.count("C")

print("A:", count_A)
print("T:", count_T)
print("G:", count_G)
print("C:", count_C)
```

This code counts the frequency of each nucleotide in a DNA sequence.
This is a basic operation in genomic data analysis.
