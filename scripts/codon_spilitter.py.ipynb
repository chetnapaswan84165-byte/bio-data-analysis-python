```python
# Input - DNA sequence 
# Process - Convert DNA to RNA, Read RNA in groups of 3 nuclotides, Ignore incomplete triplet at the end or warn
# Output - RNA sequence and RNA sequence spilit in codon(triplets) and codon print line by line.
```

```python
def dna_to_rna(dna):
    rna = ""
    for nucleotide in dna:
        if nucleotide == "A":
            rna += "U"
        elif nucleotide == "T":
            rna += "A"
        elif nucleotide == "G":
            rna += "C"
        elif nucleotide == "C":
            rna += "G"
    return rna

def spilit_into_codons(rna):
    codons = []
    for i in range (0, len(rna), 3):
        codon = rna[i:i+3]
        if len(codon) == 3:
            codons.append(codon)
        else:
            return "RNA sequence mutated."   
    return codons 

dna_sequence = input("Enter DNA sequence: ")
rna_sequence = dna_to_rna(dna_sequence)

print("RNA sequence:", rna_sequence)

codons = spilit_into_codons(rna_sequence)
print("Codons:")
for  c  in  codons:
    print(c)
```
