```python
import random

def dna_mutation(dna):
    dna_list = list(dna)
    position = random.randint(0, len(dna_list) - 1)
    
    original = dna_list[position]
    nucleotides = ["A","T","G","C"]
    nucleotides.remove(original)
    
    dna_list[position] = random.choice(nucleotides)
    mutated_dna = "".join(dna_list)
    return position, original, mutated_dna

dna_sequence = input("Enter DNA sequence:")
pos, original, mutated = dna_mutation(dna_sequence)

print("Original sequence:", dna_sequence)
print("Mutated sequence:", mutated)
print("Mutation position:", pos)
print("Original nucleotide:",original)
```
