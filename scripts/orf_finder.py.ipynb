```python
rna_sequence = input("Enter RNA sequence:").upper()

codon_table = {"AUG":"M", "UUU":"F", "UUC":"F", "UUA": "L", "UUG":"L", "ACC":"T", "ACU":"T", "ACA":"T", "ACG":"T", "UGG":"W","UAA":"STOP", "UAG":"STOP", "UGA":"STOP"}


import random
random.seed(42)

def spilit_codons_by_frame(rna, frame):
    codons = []
    for i in range(frame, len(rna), 3):
        codon = rna[i:i+3]
        if len(codon) == 3:
            codons.append(codon)
    return codons


def find_orfs(rna, codon_table):
    orfs = []
    
    for frame in range(3):
        i = frame
        while i<= len(rna)-3:
            codon = rna[i:i+3]
            
            if codon == "AUG":
                protein = ""
                start = i
                
                j = i
                while j<= len(rna) -3:
                    current_codon = rna[j:j+3]
                    amino_acid = codon_table.get(current_codon)
                    
                    if amino_acid == "STOP":
                        orfs.append({"frame": frame + 1, "start": start, "stop": j + 2, "protein": protein})
                        break
                        
                    if amino_acid:
                        protein += amino_acid
                        
                    j += 3
                    
                i = j
            else:
                i += 3
             
    return orfs  

orfs = find_orfs(rna_sequence, codon_table)

if not orfs:
    print("No ORFs found")
    
for orf in orfs:
    print("\nORF found")
    print("Frame:", orf["frame"])
    print("Start:",orf["start"])
    print("Stop:", orf["stop"])
    print("Protein:", orf["protein"])





def rna_mutation(rna):
    rna_sequence = list(rna)
    position = random.randint(0, len(rna_sequence) - 1)
    
    original = rna_sequence[position]
    nucleotides = ["A","U","G","C"]
    nucleotides.remove(original)
    
    rna_sequence[position] = random.choice(nucleotides)
    mutated_rna = "".join(rna_sequence)
    return position, original, mutated_rna

pos, original, mutated_rna = rna_mutation(rna_sequence)
print("Mutation position:", pos)
print("Original nucleotide:", original)
print("Mutated RNA:", mutated_rna) 



def find_mutated_orfs(mutated_rna, codon_table):
    mutated_orfs = []
    
    for frame in range(3):
        i = frame
        while i<= len(mutated_rna)-3:
            codon = mutated_rna[i:i+3]
            
            if codon == "AUG":
                protein = ""
                start = i
                
                j = i
                while j<= len(mutated_rna) -3:
                    current_codon = mutated_rna[j:j+3]
                    amino_acid = codon_table.get(current_codon)
                    
                    if amino_acid == "STOP":
                        mutated_orfs.append({"frame": frame + 1, "start": start, "stop": j + 2, "protein": protein})
                        break
                    
                        
                    if amino_acid:
                        protein += amino_acid
                        
                    j += 3
                    
                i = j
            else:
                i += 3
             
    return mutated_orfs 


mutated_orfs = find_mutated_orfs(mutated_rna, codon_table)
print("Mutated ORFs", mutated_orfs)
    

    
    

for mutated_orf in mutated_orfs:
    print("\nORF found")
    print("Frame:", mutated_orf["frame"])
    print("Start:",mutated_orf["start"])
    print("Stop:", mutated_orf["stop"])
    print("Protein:", mutated_orf["protein"])
```
