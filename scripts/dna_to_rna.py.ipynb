# DNA to RNA (Transcription)

```python
def dna_to_rna(dna):
    rna = ""
    for nucleotide in dna:
        if nucleotide == "A":
            rna += "U"
        elif nucleotide == "T":
            rna += "A"
        elif nucleotide == "G":
            rna += "C"
        elif nucleotide == "C":
            rna += "G"
    return rna

dna_sequence = input("Enter DNA sequence:")
print(dna_to_rna(dna_sequence))
```
