```python
AA_ONE = {
    "AUG": "M",
    "UUU": "F", "UUC": "F",
    "UUA": "L", "UUG": "L",
    "ACC": "T", "ACU": "T", "ACA": "T", "ACG": "T",
    "UGG": "W",
    "UAA": "*", "UAG": "*", "UGA": "*"
}

def find_orfs(rna, codon_map):
    assert set(rna).issubset({"A", "U", "G", "C"}), "Invalid RNA"

    orfs = []

    for frame in range(3):
        i = frame
        while i <= len(rna) - 3:
            codon = rna[i:i+3]

            if codon == "AUG":
                protein = ""
                start = i
                j = i

                while j <= len(rna) - 3:
                    aa = codon_map.get(rna[j:j+3])

                    if aa == "*":
                        orfs.append({
                            "frame": frame + 1,
                            "start": start,
                            "stop": j + 2,
                            "protein": protein,
                            "length": len(protein)
                        })
                        break

                    if aa:
                        protein += aa

                    j += 3

                i = j
            else:
                i += 3

    return orfs

rna = "AUGUUUUAA"
orfs = find_orfs(rna, AA_ONE)

orf = orfs[0]

#start codon validation
assert rna[orf["start"]: orf["start"]+3] == "AUG"

#stop codon validation
assert rna[orf["stop"]-2: orf["stop"]+1] in {"UAA", "UAG", "UGA"}

#length consistency
nt_length = orf["stop"] - orf["start"] + 1
assert nt_length % 3 == 0
assert len(orf["protein"]) == (nt_length // 3) - 1

#Frame consistency
assert orf["start"] % 3 == (orf["frame"] - 1)
```

# ORF scoring system 

```python
AA_ONE = {
    "AUG": "M",
    "UUU": "F", "UUC": "F",
    "UUA": "L", "UUG": "L",
    "ACC": "T", "ACU": "T", "ACA": "T", "ACG": "T",
    "UGG": "W",
    "GGG": "G", "GGA":"G", "GGC": "G", "GGU":"G",
    "CCU": "P", "CCC" : "P", "CCA" : "P", "CCG" : "P",
    "UAA": "*", "UAG": "*", "UGA": "*"
}


def gc_content(seq):
    return (seq.count("G") + seq.count("C")) / len(seq)





def find_orfs(rna, codon_map):
    assert set(rna).issubset({"A", "U", "G", "C"}), "Invalid RNA"

    orfs = []

    for frame in range(3):
        i = frame
        while i <= len(rna) - 3:
            codon = rna[i:i+3]
            frame_penalty = 0 if frame == 0 else 1

            if codon == "AUG":
                protein = ""
                start = i
                j = i

                while j <= len(rna) - 3:
                    aa = codon_map.get(rna[j:j+3])

                    if aa == "*":
                        nt_length = (j + 3) - start
                        orf_rna = rna[start:j+3]
                        score = (len(protein) * 2) + (gc_content(orf_rna) * 10) - (frame_penalty * 5)
                        orfs.append({
                            "frame": frame + 1,
                            "start": start,
                            "stop": j + 2,
                            "protein": protein,
                            "nt_length": nt_length,
                            "gc_content": gc_content(orf_rna),
                            "score": score
                        })
                        break

                    if aa:
                        protein += aa

                    j += 3

                i = j
            else:
                i += 3

    return orfs





rna = "AUGGGGCCCUAA"
orfs = find_orfs(rna, AA_ONE)

best = sorted(orfs, key=lambda x: x["score"], reverse=True)[0]

assert best["protein"] == "MGP"
assert best["gc_content"] > 0.5
```
