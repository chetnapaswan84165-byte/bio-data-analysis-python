```python
# Input DNA sequence
# Output convert to RNA
#Output generate codons in frames  and print codons for each frame clearly.
```

```python
def dna_to_rna(dna):
    rna = ""
    for nucleotide in dna:
        if nucleotide  == "A":
            rna += "U"
         
        elif nucleotide == "T":
            rna += "A"
        elif nucleotide == "G":
            rna += "C"
        elif nucleotide == "C":
            rna += "G"
    return rna

def get_codons_in_frame(rna, start):
    codons = []
    for i in range(start, len(rna), 3):
        codon = rna[i:i+3]
        if len(codon) == 3:
            codons.append(codon)
    return codons

dna_sequence = input("Enter DNA sequence:")
rna_sequence = dna_to_rna(dna_sequence)

print("RNA sequence:", rna_sequence)

for frame in range(3):
    print(f"\nFrame{frame +1}:")
    codons = get_codons_in_frame(rna_sequence, frame)
    for c in codons:
        print(c)
```
