# Compare ORFs across multiple sequences(simulate conservation)
# Process: Taking both sequences of RNA and performing function on it to find ORF.
# Output: Compare for both RNA sequences Number of ORFs, Longest protein length, presence/absence of specific ORFs 

```python
AA_ONE = {"AUG":"M", "UUU":"F", "UUC" :"F", "UUA" :"L", "UUG" : "L", "ACC" : "T", "ACU" : "T", "ACG" :"T", "ACA" : "T", "UGG":"W","UAA":"*", "UAG": "*", "UGA":"*"}

rna1 = "AUGGGGCCCAUCACAUUCUUGAUGACCUGGUAA"
rna2 = "AUGUUCUUGACCUGGAUGGGGAUGACUCCCUACUAG"  #single nucleotie change to simulate species difference


def find_orfs(rna, codon_map):
    orfs = []
    for frame in range (3):
        i = frame 
        while i<= len(rna) - 3:
            if rna[i:i+3] == "AUG":
                protein = ""
                start = i
                j=i
                while j<= len(rna)-3:
                    codon = rna[j:j+3]
                    aa = codon_map.get(codon)
                  
                    
                    if aa == "*":
                        orfs.append({"frame": frame + 1, "start": start, "stop": j + 2, "protein" : protein})
                        break
                    if aa:
                        protein += aa
                        
                    j += 3
                i = j +3
            else:
                i += 3
    return orfs


print("\n----COMPARISON REPORT----")

print("RNA Sequence 1", rna1)
print("RNA Sequence 2", rna2)
orfs = find_orfs(rna1, AA_ONE)
print("ORFs of rna1 sequence", orfs)
orfs = find_orfs(rna2, AA_ONE)
print("ORFs of rna2 sequence", orfs)
orfs1 = find_orfs(rna1, AA_ONE)
orfs2 = find_orfs(rna2, AA_ONE)


print("Protein from RNA1:")
for o in orfs1:
    print(o["protein"])
print("Protein from RNA2:")
for o in orfs2:
    print(o["protein"])
    
def longest_orf(orfs):
    if not orfs:
        return None
    return max(orfs, key=lambda x: len(x["protein"]))

l1 = longest_orf(orfs1)
l2 = longest_orf(orfs2)

print("Longest protein RNA1:", l1["protein"] if l1 else None)
print("Longest protein RNA2:", l2["protein"] if l2 else None)

def longest_protein(orfs):
    if not orfs:
        return 0
    return max(len(o["protein"]) for o in orfs)
print("Longest ORF RNA1:", longest_protein(orfs1))
print("Longest ORF RNA2:", longest_protein(orfs2))


    
    

```
