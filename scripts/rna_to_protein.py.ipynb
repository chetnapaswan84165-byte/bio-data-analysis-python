```python
# Input :- RNA sequence
# Process:- Start translation at first AUG, 
# Translate codns using a codon table, Stop translation at first codon.
# Output :- amino acid sequence.
```

```python
codon_table = {"AUG":"(M)", "UUU":"(F)", "UUC":"(F)","UUA": "(L)", "UUG":"(L)", "UAA":"STOP","UAG":"STOP","UGA":"STOP"}

def rna_to_protein(rna):
    rna = rna.upper()
    protein = ""
    start_index = rna.find("AUG")
    if start_index == -1:
        return "No start codon found"
    
    for i in range(start_index, len(rna), 3):
        codon=rna[i:i+3]
        if len(codon) < 3:
            break
            
        amino_acid = codon_table.get(codon)
        if amino_acid == "STOP":
            break
        if amino_acid is None:
            continue
            
        protein+= amino_acid
    return protein

def spilit_into_codons(rna, start=0):
    codons = []
    for i in range(start,len(rna),3):
        codon = rna[i:i+3]
        if len(codon) == 3:
            codons.append(codon)
    return codons 

rna_sequence = input("Enter RNA sequence:")
start = rna_sequence.find("AUG")
if start != -1:
    print("Codons:")
    for c in spilit_into_codons(rna_sequence, start):
        print(c)
        
        
print("Protein:", rna_to_protein(rna_sequence))
```

```python
# Generate codons for frame 0, 1, and 2
# Translate each frame independently
# Clearly print the result for each frame
```

```python
codon_table = {"AUG": "Met(M)", "UUU":"Phe(F)", "UUC": "Phe(F)", "UUA":"Leu(L)", "UUG": "Leu(L)", "UAA":"STOP", "UGA": "STOP", "UAG":"STOP"}

def spilit_codons_by_frame(rna, frame):
    codons = []
    for i in range(frame, len(rna), 3):
        codon = rna[i:i+3]
        if len(codon) == 3:
            codons.append(codon)
     
    return codons

def translate_codons(codons, codon_table):
    protein = ""
    for codon in codons:
        amino_acid = codon_table.get(codon)
        if amino_acid == "STOP":
            break
        if amino_acid:
            protein += amino_acid
     
    return protein       

rna_sequence = input("Enter RNA sequence:").upper()

for frame in range(3):
    codons = spilit_codons_by_frame(rna_sequence, frame)
    protein = translate_codons(codons, codon_table)
    
    print(f"\nFrame {frame +1}")
    print("Codons:", codons)
    print("Protein:", protein if protein else "No valid protein")
    
```
